### TASK  3: Visualize the relationship between success rate of each orbit type


Next, we want to visually check if there are any relationship between success rate and orbit type.


Let's create a `bar chart` for the sucess rate of each orbit


```python
# HINT use groupby method on Orbit column and get the mean of Class column

orbit_success = df.groupby('Orbit')['Class'].mean().sort_values(ascending=0)

sns.barplot (
    x=orbit_success.index,
    y=orbit_success.values
)

plt.xlabel('Orbit')
plt.ylabel('Success rate')
plt.show()
```

Analyze the ploted bar chart try to find which orbits have high sucess rate.


#### Observation
- ES-L1, GEO, HEO, SSO and VLEO show the highest landing success rates.
- Success rate should be interpreted together with the number of launches. A 100% success rate based on one launch is much less reliable than a similar rate based on many launches.
- From this visualization alone, we cannot conclude how orbit type independently affects landing success.

```python
# HINT use groupby method on Orbit column and get the mean of Class column

orbit_success = df.groupby('Orbit')['Class'].agg(['mean', 'count']).sort_values(by='count', ascending=0)
orbit_success
```

#### Additional Observation

- VLEO stands out with a high landing success rate of about 86% across 14 launches, making this result more convincing than 100% success rates based on only 1–5 launches.
- GTO and ISS have the largest number of launches, but noticeably lower landing success rates.

### TASK  4: Visualize the relationship between FlightNumber and Orbit type


For each orbit, we want to see if there is any relationship between FlightNumber and Orbit type.


```python
# Plot a scatter point chart with x axis to be FlightNumber and y axis to be the Orbit, and hue to be the class value

sns.catplot (
    x='FlightNumber',
    y='Orbit',
    hue='Class',
    data = df
)

plt.xlabel('FlightNumber')
plt.ylabel('Orbit')
plt.show()
```

#### Observation

- LEO, ISS, PO and GTO appear among the earliest missions.
- VLEO appears only in the later part of the launch history and becomes frequent there.
- VLEO missions in this dataset show a high landing success rate.
- For some orbit types, especially GTO, later flights appear more successful than earlier flights.
- This suggests that FlightNumber may reflect accumulated operational experience and technological development, so the effect of orbit type alone is difficult to isolate from this plot.

You should see that in the LEO orbit the Success appears related to the number of flights; on the other hand, there seems to be no relationship between flight number when in GTO orbit.
